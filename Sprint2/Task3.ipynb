### Вариант №11

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Customer в C#, который будет представлять информацию о 
клиентах или покупателях. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов  должны  быть  реализованы  новые  атрибуты  и  методы,  а  также 
переопределены  некоторые  методы  базового  класса  для  демонстрации 
полиморфизма

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
// interfaces (multiple inheritance)
interface IEmailable { void SendEmail(string subj); }
interface IDiscountable { decimal GetDiscount(decimal amount); }

// base class
abstract class Customer : IEmailable
{
    public int CustomerId { get; set; }
    public string Name { get; set; } = "";
    public string Email { get; set; } = "";
    public bool IsActive { get; private set; } = true;
    public DateTime CreatedAt { get; init; } = DateTime.Now;

    public virtual string GetFullName() => Name;

    public virtual void UpdateEmail(string newEmail)
    {
        Email = newEmail;
        Console.WriteLine($"Email обновлён: {Email}");
    }

    public virtual void ViewProfile() =>
        Console.WriteLine($"ID: {CustomerId}, Имя: {Name}, Email: {Email}, Активен: {IsActive}");

    public void Deactivate() => IsActive = false;
    public void SendEmail(string subj) => Console.WriteLine($"Отправлено письмо: {subj} -> {Email}");
}

// simple inheritance
class VipCustomer : Customer, IDiscountable
{
    public int Points { get; set; }
    public string Tier { get; set; } = "Gold";
    public decimal DiscountPercent { get; set; } = 10m;

    public void AddPoints(int p) {
        Points += p;
        Console.WriteLine($"Добавлено {p} баллов вип пользователю");
        }
    public decimal GetDiscount(decimal amount) => amount * (DiscountPercent / 100m);

    public override void ViewProfile() =>
        Console.WriteLine($"[VIP] {Name}, Баллы: {Points}, Скидка: {DiscountPercent}%");
}

// another child of customer
class RegularCustomer : Customer
{
    public DateTime Registered { get; set; }
    public int Orders { get; private set; }

    public void AddOrder() {
        Orders++;
        Console.WriteLine("Добавлен заказ");
        }
    public override void ViewProfile() =>
        Console.WriteLine($"[REG] {Name}, заказов: {Orders}, зарегистрирован: {Registered:d}");
}

// multilevel(сложное) inheritance
class CorporateCustomer : RegularCustomer, IDiscountable
{
    public string CompanyName { get; set; } = "";
    public decimal CreditLimit { get; set; } = 50000m;

    public void SpendCredit(decimal sum) {
        CreditLimit -= sum;
        Console.WriteLine($"Потрачено {sum} кредитов");
        }
    public decimal GetDiscount(decimal amount) => amount * 0.15m;

    public override void ViewProfile() =>
        Console.WriteLine($"[CORP] {CompanyName}, Кредит: {CreditLimit}");
}


// VipCustomer : Customer, IDiscountable — простое + множественное (интерфейс)
Customer vip = new VipCustomer { CustomerId = 1, Name = "Никита", Email = "nik@mail.com", Points = 120 };

// RegularCustomer : Customer — простое наследование
Customer reg = new RegularCustomer { CustomerId = 2, Name = "Егор", Email = "egor@mail.com", Registered = new DateTime(2023, 1, 1) };

// CorporateCustomer : RegularCustomer : Customer, IDiscountable — многоуровневое + множественное
Customer corp = new CorporateCustomer { CustomerId = 3, Name = "ACME Contact", Email = "corp@acme.com", CompanyName = "Acme Ltd.", CreditLimit = 50000m };
((CorporateCustomer)corp).Registered = new DateTime(2022, 10, 1); 

Console.WriteLine("Вывод созданных пользователей");
vip.ViewProfile();   // версия VipCustomer
reg.ViewProfile();   // версия RegularCustomer
corp.ViewProfile();  // версия CorporateCustomer

Console.WriteLine("\nДоп методы потомков");
if (vip is VipCustomer v) { v.AddPoints(30); Console.WriteLine($"VIP баллы: {v.Points}"); }
if (reg is RegularCustomer r) { r.AddOrder(); Console.WriteLine($"REG заказы: {r.Orders}"); }
if (corp is CorporateCustomer c) { c.SpendCredit(2000m); Console.WriteLine($"CORP кредит: {c.CreditLimit}"); }

Console.WriteLine("\nМножественное наследование через интерфейс IDiscountable");
Console.WriteLine("Скидки для различных профилей:");
if (vip is IDiscountable dv) Console.WriteLine($"VIP скидка от 1000: {dv.GetDiscount(1000m)}");
if (reg is IDiscountable) Console.WriteLine("REG скидка: нет (интерфейс не реализован)");
if (corp is IDiscountable dc) Console.WriteLine($"CORP скидка от 1000: {dc.GetDiscount(1000m)}");

Console.WriteLine("\nБазовый Customer, использующий интерфейс IEmailable");
Console.WriteLine("Отправка email:");
vip.SendEmail("Добро пожаловать, VIP!");
reg.SendEmail("Подтвердите e-mail");
corp.SendEmail("Ваш корпоративный менеджер");

Console.WriteLine("\nПросмотр всех профилей");
var list = new List<Customer> { vip, reg, corp };
foreach (var cst in list) cst.ViewProfile(); // один вызов — разные реализации

```

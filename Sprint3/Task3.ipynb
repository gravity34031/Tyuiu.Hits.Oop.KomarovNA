Создать базовый класс Customer в C#, который будет представлять информацию о 
клиентах или покупателях. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов  должны  быть  реализованы  новые  атрибуты  и  методы,  а  также 
переопределены  некоторые  методы  базового  класса  для  демонстрации 
полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

<h2 style="color:DodgerBlue">Реализация:</h2>

```
abstract class Customer
{
    public int CustomerId { get; set; }
    public string Name { get; set; }
    public string Email { get; set; }
    public string Phone { get; set; }           // новый атрибут
    public bool IsActive { get; set; } = true;  // новый атрибут

    public virtual string GetFullName()
    {
        return Name;
    }

    public virtual void UpdateEmail(string newEmail)
    {
        Email = newEmail;
    }

    // Перегрузка (overload) метода UpdateEmail — с логом изменений
    public void UpdateEmail(string newEmail, bool showLog)
    {
        Email = newEmail;
        if (showLog)
            Console.WriteLine($"[LOG] Email для {Name} изменён на {newEmail}");
    }

    public virtual void Deactivate()
    {
        IsActive = false;
    }

    public virtual void ViewProfile()
    {
        Console.WriteLine($"ID: {CustomerId}\nИмя: {Name}\nEmail: {Email}\nТелефон: {Phone}\nСтатус: {(IsActive ? "Активен" : "Неактивен")}");
    }
}


class VipCustomer : Customer
{
    public int LoyaltyPoints { get; set; }
    public decimal DiscountRate { get; set; }   // новый атрибут
    public DateTime VipStartDate { get; set; }  // новый атрибут

    public override void ViewProfile()
    {
        Console.WriteLine($"--- VIP КЛИЕНТ ---");
        Console.WriteLine($"ID: {CustomerId}\nИмя: {Name}\nEmail: {Email}\nТелефон: {Phone}\nБаллы: {LoyaltyPoints}\nСкидка: {DiscountRate}%\nVIP с: {VipStartDate:d}");
        Console.WriteLine();
    }

    public override void Deactivate()
    {
        Console.WriteLine($"❌ VIP-клиент {Name} был деактивирован, баллы сохранены ({LoyaltyPoints}).");
        base.Deactivate();
    }
}

class RegularCustomer : Customer
{
    public DateTime RegistrationDate { get; set; }
    public DateTime LastEmailUpdate { get; private set; }
    public int OrdersCount { get; set; }  // новый атрибут

    public override void UpdateEmail(string newEmail)
    {
        base.UpdateEmail(newEmail);
        LastEmailUpdate = DateTime.Now;
    }

    public override void ViewProfile()
    {
        Console.WriteLine($"--- Обычный клиент ---");
        Console.WriteLine($"ID: {CustomerId}\nИмя: {Name}\nEmail: {Email}\nТелефон: {Phone}\nЗаказы: {OrdersCount}\nЗарегистрирован: {RegistrationDate:d}\nПоследнее обновление Email: {LastEmailUpdate:g}");
        Console.WriteLine();
    }
}

class GroupCustomer : Customer
{
    public string GroupName { get; set; }
    public int MembersCount { get; set; } // новый атрибут

    public override string GetFullName()
    {
        return $"Группа \"{GroupName}\"";
    }

    public override void ViewProfile()
    {
        Console.WriteLine($"--- Групповой клиент ---");
        Console.WriteLine($"ID: {CustomerId}\nГруппа: {GroupName}\nEmail: {Email}\nТелефон: {Phone}\nКоличество участников: {MembersCount}");
        Console.WriteLine();
    }
}

// GENERIC КЛАСС
class CustomerCollection<T> where T : Customer
{
    private List<T> _customers = new List<T>();

    public void Add(T customer)
    {
        _customers.Add(customer);
    }

    public void ShowAll()
    {
        Console.WriteLine("\n📋 Список клиентов: (реализовано через generic)\n");
        foreach (var c in _customers)
        {
            c.ViewProfile();
        }
    }
}




Console.OutputEncoding = System.Text.Encoding.UTF8;
Console.WriteLine("=== СИСТЕМА УПРАВЛЕНИЯ КЛИЕНТАМИ ===\n");

var vip = new VipCustomer
{
    CustomerId = 1,
    Name = "Никита",
    Email = "nikita@mail.com",
    Phone = "+7 999 000-11-22",
    LoyaltyPoints = 120,
    DiscountRate = 15,
    VipStartDate = new DateTime(2023, 10, 1)
};

var regular = new RegularCustomer
{
    CustomerId = 2,
    Name = "Егор",
    Email = "egor@mail.com",
    Phone = "+7 900 123-45-67",
    RegistrationDate = new DateTime(2022, 5, 1),
    OrdersCount = 5
};

var group = new GroupCustomer
{
    CustomerId = 3,
    GroupName = "Студенты",
    Email = "group@mail.com",
    Phone = "+7 999 888-77-66",
    MembersCount = 12
};

// Добавляем клиентов в generic-коллекцию
var collection = new CustomerCollection<Customer>();
collection.Add(vip);
collection.Add(regular);
collection.Add(group);

// Обновляем email и показываем всех клиентов
regular.UpdateEmail("egor.new@mail.com", true);
vip.Deactivate();

collection.ShowAll();

Console.WriteLine("=== Конец программы ===");
```
